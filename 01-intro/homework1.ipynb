#### Q7. Sum of weights

```python
# 1. all the cars from Asia

df_asia_cars = df[
    df['origin'] == 'Asia'
]
df_asia_cars.head()
```

```python
# 2 & 3. only columns vehicle_weight and model_year and the first 7 values

X_df = df_asia_cars[['vehicle_weight', 'model_year']].head(7)
X_df
```

```python
# 4. the underlying NumPy array

X = X_df.values
X
```

```python
# 5. the transpose of X
XT = X.T
XT
```

```python
# 5. multiplication between the transpose of X and X

XTX = XT.dot(X)
XTX
```

```python
# 6. Invert XTX

XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# 7. New array
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# 8
r = XTX_inv.dot(XT)
w = r.dot(y)
w
```

```python
w.sum()
```
